```python
# 간단한 버전
import numpy as np
import matplotlib.pyplot as plt
import cv2

def simple_heatmap_diff(image1_path, image2_path):
    """
    간단한 히트맵 차이 분석
    """
    # 이미지 불러오기
    img1 = cv2.imread(image1_path, cv2.IMREAD_GRAYSCALE) / 255.0
    img2 = cv2.imread(image2_path, cv2.IMREAD_GRAYSCALE) / 255.0
    
    # 크기 통일
    if img1.shape != img2.shape:
        img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
    
    # 차이 계산 (앞에서 뒤를 뺌)
    diff = img1 - img2
    
    # 시각화
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    axes[0].imshow(img1, cmap='viridis')
    axes[0].set_title('이미지1 (앞)')
    axes[0].axis('off')
    
    axes[1].imshow(img2, cmap='viridis')
    axes[1].set_title('이미지2 (뒤)')
    axes[1].axis('off')
    
    im = axes[2].imshow(diff, cmap='jet', vmin=-np.abs(diff).max(), vmax=np.abs(diff).max())
    axes[2].set_title('차이 (앞 - 뒤)')
    axes[2].axis('off')
    plt.colorbar(im, ax=axes[2])
    
    plt.tight_layout()
    plt.show()
    
    print(f"차이 통계: 최소={diff.min():.3f}, 최대={diff.max():.3f}, 평균={diff.mean():.3f}")
    
    return diff

# 사용법
# simple_heatmap_diff("image1.png", "image2.png")
```
